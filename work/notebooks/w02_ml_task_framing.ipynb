## 2. Target or proxy
**Target / proxy**

The target is a proxy for refresh opportunity based on observable page-level search and content signals. The starter data provides an observed trend direction, including whether a page is declining. I will use this as a measurable proxy for prioritization rather than claiming that declining status directly proves that a refresh is needed. The target comes from an observed outcome in the dataset, not from a product decision flag.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

csv_url = "https://raw.githubusercontent.com/Srinadh2314/srinadh-flyrank-intership/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(csv_url)

df["is_declining"] = df["trend_direction"].str.lower().eq("down").astype(int)

print("Rows:", len(df))
print("Declining pages:", df["is_declining"].sum())
print("Declining rate:", round(df["is_declining"].mean(), 3))
```

## 3. Success metric
**Success metric: Precision@50**

I will use Precision@50 because the practical decision is which pages to review first when review time is limited. Precision@50 measures the proportion of the top 50 ranked pages that match the target definition. A higher value means the highest-priority part of the ranking contains more pages with the target signal. This measures ranking usefulness, not whether a refresh will definitely improve a page.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
def precision_at_k(scores, labels, k):
    order = scores.argsort()[::-1][:k]
    return labels.iloc[order].mean()

print("Example target prevalence:", round(df["is_declining"].mean(), 3))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A page can be old, highly visible, have a different average position, and have a different CTR pattern at the same time. ML can learn how multiple observable signals interact and use them together to produce a ranking. The benefit I am looking for is better prioritization of limited review time, not an automatic decision about what action a page needs.
